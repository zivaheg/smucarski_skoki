```python
correlation_matrix_A = df_A.iloc[:, 1:].corr()
correlation_matrix_B = df_B.iloc[:, 1:].corr()

strong_correlations_A = correlation_matrix_A.unstack().sort_values(ascending=False)
strong_correlations_B = correlation_matrix_B.unstack().sort_values(ascending=False)

strong_correlations_A = strong_correlations_A[strong_correlations_A < 1]
strong_correlations_B = strong_correlations_B[strong_correlations_B < 1]

strong_correlations_A
```

## Here are some notable correlations from the two matrices:

### Matrix A (Physical Features - Ski Jump Simulation):

Vy [m/s] and Roll Angle Left [°]: 0.257 – This suggests a moderate relationship between the vertical velocity component and the roll angle to the left.

Vy [m/s] and Roll Angle Right [°]: 0.210 – The vertical velocity also correlates moderately with the roll angle to the right.

Opening Angle [°] and Speed [m/s]: 0.157 – There's a mild positive correlation between the opening angle of the ski jump and the speed.

Vx [m/s] and Yaw Angle Right [°]: 0.127 – The horizontal velocity (Vx) seems somewhat related to the yaw angle.

Stalling Angle Right [°] and Speed [m/s]: 0.098 – A weak positive correlation between the stalling angle and speed.

### Matrix B (Trajectory Metrics - Ski Jump):

landing_Cross_quad and landing_Tangent_quad: 0.94 – Extremely strong correlation, indicating that landing cross and tangent angles are highly dependent on each other.

landing_Tangent and landing_Cross_quad: 0.89 – A high correlation between landing tangent and the cross component.

mid_Tangent and takeoff_Speed: 0.87 – Strong correlation, suggesting a link between the tangent angle at the middle of the jump and the takeoff speed.

takeoff_Tangent and mid_Cross_quad: 0.84 – A high correlation between the takeoff tangent and the middle cross component.

### Physical Insights:

In Matrix A, the correlations between the vertical velocity components (Vy) and roll angles (both left and right) suggest that the skier's body orientation (roll) might influence or be influenced by the vertical velocity, which is key in understanding the flight dynamics and the stability of the jump.

Matrix B indicates that the landing components (cross and tangent) have very high correlations, which makes sense because these two components are critical in determining the landing position and angle of the ski jump.
